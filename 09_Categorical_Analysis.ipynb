## Unique values, frequencies, rare categories, and cardinality

### 1. Concept
Cardinality is the number of distinct categories. Rare categories occur infrequently; high-cardinality columns contain many distinct labels.

### 2. Simple example
A country field has fewer unique values than a unique transaction ID.

### 3. Business example
Inspect product types, store names, or statuses.

### 4. AI/ML relevance
Review passenger sex, class, and embarkation categories.

```python
Rare and high-cardinality categories can affect encoding and model stability.
```

### 5. Output explanation
rows=[]
for col in df.select_dtypes(exclude=np.number).columns:
    counts=df[col].value_counts(dropna=False)
    rows.append({"column":col,"unique_including_missing":df[col].nunique(dropna=False),"rare_categories_count_le_5":int((counts<=5).sum()),"missing":int(df[col].isna().sum())})
display(pd.DataFrame(rows).sort_values("unique_including_missing",ascending=False))

### 6. Interpretation / insight
Category handling depends on domain meaning.

### 7. Best practices

## Categorical features versus target

### 1. Concept
Cross-tabulation and normalized proportions show how a categorical variable relates to a target class.

### 2. Simple example
Compare outcomes across customer segments.

### 3. Business example
Compare return rates by product category.

### 4. AI/ML relevance
Compare observed survival proportions across passenger groups.

```python
Provides exploratory hypotheses while allowing for confounding.
```

### 5. Output explanation
if "survived" in df:
    for col in [c for c in ["sex","pclass","embarked","alone"] if c in df]:
        display(pd.crosstab(df[col],df["survived"],normalize="index").round(3))
        pd.crosstab(df[col],df["survived"],normalize="index").plot(kind="bar",stacked=True,figsize=(7,3))
        plt.title("Target proportions by "+col); plt.ylabel("Proportion"); plt.tight_layout(); plt.show()

### 6. Interpretation / insight
Associations can be confounded by other variables.

### 7. Best practices
Normalized tables compare proportions, not just raw group sizes.

### 8. Common mistakes
Include group sizes and avoid causal claims.

### 9. Advantages
Using raw counts to compare groups of different sizes.

### 10. Limitations
Easy to interpret.

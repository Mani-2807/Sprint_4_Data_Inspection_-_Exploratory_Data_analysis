# Notebook 12 — Business Insights
Convert EDA outputs into observations, interpretations, caveats, and next steps. A chart description alone is not an insight.

## Insight framework
1. **Observation:** What did the data show?
2. **Evidence:** Which statistic, count, percentage, or plot supports it?
3. **Interpretation:** What might it mean?
4. **Caveat:** What else could explain it?
5. **Next step:** What should be investigated next?

```python
evidence=[]
if "survived" in df:
    evidence.append(f"Overall observed survival rate: {df['survived'].mean()*100:.1f}% of non-missing target values.")
for col in ["sex","pclass","embarked"]:
    if col in df and "survived" in df:
        rates=(df.groupby(col,dropna=False)["survived"].mean()*100).round(1)
        evidence.append(f"Observed survival percentages by {col}: {rates.to_dict()}")
if "fare" in df:
    evidence.append(f"Fare median: {df['fare'].median():.2f}; mean: {df['fare'].mean():.2f}.")
for e in evidence:
    print("-",e)
```

```python
%pip install -q pandas numpy matplotlib seaborn
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme()

DATA_URL = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(DATA_URL)
print("Loaded shape:", df.shape)
display(df.head())
```

# Notebook 13 — EDA Report
This notebook builds a concise report from computed outputs. It identifies issues and recommends next steps; it does not complete preprocessing.

## 1. Dataset overview

```python
overview=pd.Series({
    "name":"Titanic passenger dataset (Seaborn repository)",
    "source":DATA_URL,
    "domain":"Historical passenger survival / classification",
    "rows":int(df.shape[0]),
    "columns":int(df.shape[1]),
    "target":"survived" if "survived" in df else None
},name="value")
display(overview.to_frame())
display(pd.DataFrame({"dtype":df.dtypes.astype(str),"non_null":df.notna().sum(),"missing":df.isna().sum(),"missing_pct":(df.isna().mean()*100).round(2)}))
```

## 2. Data quality

```python
display(pd.DataFrame({
    "missing_count":df.isna().sum(),
    "missing_percent":(df.isna().mean()*100).round(2),
    "unique_including_missing":df.nunique(dropna=False),
    "dtype":df.dtypes.astype(str)
}).sort_values("missing_percent",ascending=False))
print("Fully duplicated rows:",int(df.duplicated().sum()))
```

## 3. Statistical findings

```python
display(df.select_dtypes(include=np.number).describe().T)
```

## 4. Relationships

```python
if "survived" in df:
    for col in [c for c in ["sex","pclass","embarked"] if c in df]:
        display(pd.crosstab(df[col],df["survived"],normalize="index").round(3))
display(df.select_dtypes(include=np.number).corr().round(2))
```

## 5. Visualization findings

```python
if "fare" in df:
    sns.histplot(data=df,x="fare",kde=True); plt.title("Fare distribution"); plt.show()
if "survived" in df and "pclass" in df:
    sns.barplot(data=df,x="pclass",y="survived",errorbar=None)
    plt.title("Observed survival proportion by passenger class"); plt.show()
```

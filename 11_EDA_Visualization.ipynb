## Histogram and distribution plot

### 1. Concept
Histograms show counts by bins; density plots estimate the distribution shape.

### 2. Simple example
Check whether values cluster or have a long tail.

### 3. Business example
Inspect the distribution of prices or order values.

### 4. AI/ML relevance
Explore fare distribution.

```python
Shape may guide later transformations.
```

### 5. Output explanation

## Box plot and violin plot

### 1. Concept
Box plots show quartiles and potential extremes; violin plots show smoothed density by group.

### 2. Simple example
Compare distributions across departments.

### 3. Business example
Compare spending by customer segment.

### 4. AI/ML relevance

```python
Can reveal group-level distribution differences.
```

### 9. Advantages
Good for numeric-by-category analysis.

### 10. Limitations
Group size and missingness affect interpretation.

## Pair plot and heatmap

### 1. Concept
Pair plots compare several feature pairs; heatmaps display a matrix such as correlations.

### 2. Simple example
A compact view of many feature relationships.

### 3. Business example
Explore correlation among business metrics.

### 4. AI/ML relevance
Inspect relationships among numeric features.

```python
Can suggest redundancy or interesting relationships.
```

### 5. Output explanation
cols=[c for c in ["age","fare","pclass","survived"] if c in df]
small=df[cols].dropna().sample(n=min(250,len(df[cols].dropna())),random_state=42)
sns.pairplot(small,hue="survived" if "survived" in small else None,corner=True); plt.show()
sns.heatmap(df.select_dtypes(include=np.number).corr(),annot=True,fmt=".2f",center=0,cmap="coolwarm")
plt.title("Correlation heatmap"); plt.show()

### 6. Interpretation / insight

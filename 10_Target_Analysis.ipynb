# Notebook 10 — Target Variable Analysis
A target is the outcome a supervised model predicts. Analyze its type, completeness, distribution, and potential quality issues before selecting models and metrics.

## Classification target distribution

### 1. Concept
Classification targets represent classes. Class imbalance occurs when some classes are much less common than others.

### 2. Simple example
A fraud dataset may contain many legitimate transactions and few fraud cases.

### 3. Business example
Understand outcome balance before choosing an evaluation metric.

### 4. AI/ML relevance
`survived` is a binary classification target in this dataset.

```python
Class balance affects splitting, baselines, and metric selection.
```

### 5. Output explanation
target="survived"
if target in df:
    counts=df[target].value_counts(dropna=False).sort_index()
    pct=(df[target].value_counts(normalize=True,dropna=False).sort_index()*100).round(2)
    display(pd.DataFrame({"count":counts,"percent":pct}))
    sns.countplot(data=df,x=target); plt.title("Target class distribution"); plt.show()
    print("Missing target:",int(df[target].isna().sum()))
else:
    print("Choose and document a valid classification target.")

### 6. Interpretation / insight
Class distribution alone does not show predictive performance.

### 7. Best practices
Counts and percentages show the target class distribution, including missing target values.

### 8. Common mistakes
Check whether a majority-class baseline could make accuracy misleading.

### 9. Advantages
Choosing metrics based on accuracy alone for imbalanced classes.

### 10. Limitations
Essential first step for classification.

## Regression target distribution

### 1. Concept
A regression target is numeric. Mean, median, variance, standard deviation, skewness, and outlier analysis describe its scale and distribution.

### 2. Simple example
Forecasting a continuous sales amount differs from predicting a class.

### 3. Business example
Inspect the distribution of sales or delivery time.

### 4. AI/ML relevance
Use fare only as an example numeric variable in this notebook, not as the primary target.

```python
Target skew and extreme values can affect loss and metric behavior.
```

### 5. Output explanation
regression_example="fare"
if regression_example in df:
    y=df[regression_example].dropna()
    print("Mean",y.mean(),"Median",y.median(),"Variance",y.var(),"Std",y.std(),"Skew",y.skew())
    display(y.describe(percentiles=[.1,.25,.5,.75,.9]).to_frame(regression_example))
    sns.histplot(y,kde=True); plt.title("Example numeric distribution"); plt.show()

### 6. Interpretation / insight
This dataset is mainly used for classification.

The code demonstrates regression-style distribution analysis on a numeric field.

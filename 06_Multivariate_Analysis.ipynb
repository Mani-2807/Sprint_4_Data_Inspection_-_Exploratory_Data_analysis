## Correlation matrix and heatmap

### 1. Concept
A correlation matrix shows pairwise numeric correlations; a heatmap encodes them visually.

### 2. Simple example
A map of which measurements tend to rise or fall together.

### 3. Business example
Check related metrics before forecasting sales.

### 4. AI/ML relevance
Inspect relationships among passenger numeric columns.

```python
Potentially redundant features can affect some model types and coefficient interpretation.
```

### 5. Output explanation
numeric=df.select_dtypes(include=np.number)
corr=numeric.corr()
display(corr.round(2))
plt.figure(figsize=(10,7)); sns.heatmap(corr,annot=True,fmt=".2f",center=0,cmap="coolwarm")
plt.title("Correlation heatmap"); plt.show()

### 6. Interpretation / insight
Correlation does not establish causation.

### 7. Best practices
Values near +1 indicate strong positive linear association; near -1 indicate negative linear association; near 0 indicates weak linear association.

### 8. Common mistakes
Check nonlinear relationships, sample sizes, and domain meaning.

### 9. Advantages
Assuming high correlation proves one feature should be removed.

### 10. Limitations
Compact summary of many pairs.

## Pair plots and feature interactions

### 1. Concept
Pair plots show pairwise relationships among selected features. An interaction occurs when a feature's relationship with an outcome depends on another feature.

### 2. Simple example
A set of small plots compares pairs of measurements.

### 3. Business example
Explore age, spending, and segment jointly.

### 4. AI/ML relevance
Check how numeric features relate to survival groups.

```python
Can suggest feature hypotheses for later validation.
```

### 5. Output explanation
cols=[c for c in ["age","fare","pclass","survived"] if c in df]
small=df[cols].dropna()
small=small.sample(n=min(250,len(small)),random_state=42)
sns.pairplot(small,hue="survived" if "survived" in small else None,corner=True)
plt.show()

### 6. Interpretation / insight
Pair plots become slow and cluttered for large datasets.

### 7. Best practices
The pair plot shows pairwise patterns and distributions in a manageable sample.

### 8. Common mistakes
Look for clusters and group-specific patterns; verify any hypothesis with further analysis.

### 9. Advantages
Using too many columns and creating unreadable plots.

### 10. Limitations
Explores several dimensions at once.

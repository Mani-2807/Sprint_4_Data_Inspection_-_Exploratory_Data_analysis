## Normal, uniform, symmetric, and skewed distributions

### 1. Concept
A normal distribution is bell-shaped and symmetric; a uniform distribution is roughly flat over a range. Right-skew has a longer right tail; left-skew has a longer left tail.

### 2. Simple example
Income data may be right-skewed when a few observations are much larger than most.

### 3. Business example
Understand whether delivery times have a long delay tail.

### 4. AI/ML relevance
Explore age and fare distributions.

```python
Distribution shape can guide later transformation choices.
```

### 5. Output explanation
for col in [c for c in ["age","fare"] if c in df]:
    s=df[col].dropna()
    print(col,"mean",round(s.mean(),2),"median",round(s.median(),2),"skew",round(s.skew(),2),"kurtosis",round(s.kurtosis(),2))
    fig,ax=plt.subplots(1,2,figsize=(9,3))
    sns.histplot(s,kde=True,ax=ax[0]); sns.boxplot(x=s,ax=ax[1])
    ax[0].set_title("Distribution: "+col); ax[1].set_title("Spread: "+col)
    plt.tight_layout(); plt.show()

### 6. Interpretation / insight
Small samples and missingness can distort distribution estimates.

### 7. Best practices
Statistics and plots describe the distribution. Pandas kurtosis reports excess kurtosis.

### 8. Common mistakes
Compare mean and median and inspect the tails rather than relying on one statistic.

### 9. Advantages
Assuming the data is normal without checking.

### 10. Limitations
Combines statistical and visual evidence.

## Percentile comparison

### 1. Concept
Percentiles show values at or below a specified percentage of observations.

### 2. Simple example
A 90th percentile is a high-end threshold for a distribution.

### 3. Business example
Set a service-level threshold based on delivery duration.

### 4. AI/ML relevance
Compare fare percentiles across passenger classes.

```python
Tail summaries can support monitoring thresholds.
```

### 5. Output explanation
if "fare" in df and "pclass" in df:
    display(df.groupby("pclass")["fare"].quantile([.1,.5,.9]).unstack().rename(columns={.1:"p10",.5:"p50",.9:"p90"}))
    sns.ecdfplot(data=df,x="fare",hue="pclass")
    plt.title("Fare cumulative distribution by class"); plt.show()

### 6. Interpretation / insight
Percentile estimates depend on sample size and method.

### 7. Best practices
Percentile tables and ECDF curves compare the distributions across groups.

### 8. Common mistakes
Check whether differences appear throughout the distribution or mainly in the tail.

### 9. Advantages
Confusing the 90th percentile with 90% of the value.

### 10. Limitations
Robust comparison of distribution tails.

# Notebook 1 — Dataset Understanding

A **dataset** is a structured collection of observations. A **feature** describes an observation; a **target** is the outcome a supervised model predicts. Independent variables are candidate predictors; the dependent variable is the outcome. Numerical features contain numbers, categorical features represent groups, discrete features are countable, and continuous features are measurements. Identifier columns label records and may not generalize as predictive signals. Metadata describes the dataset, its source, and its columns.

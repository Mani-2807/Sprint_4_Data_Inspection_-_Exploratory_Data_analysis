# Notebook 5 — Bivariate Analysis
Bivariate analysis studies relationships between two variables. Relationships do not prove causation.

## Numerical vs numerical

### 1. Concept
Scatter plots and correlation coefficients help explore associations between numeric variables.

### 2. Simple example
Compare hours studied with test scores.

### 3. Business example
Explore income versus spending or price versus quantity.

### 4. AI/ML relevance
Find possible relationships or redundant numeric features.

```python
cols=[c for c in ["age","fare","sibsp","parch"] if c in df]
if len(cols)>=2:
    sns.scatterplot(data=df,x=cols[0],y=cols[1],hue="survived" if "survived" in df else None)
    plt.title(f"{cols[0]} vs {cols[1]}"); plt.show()
    display(df[cols].corr())
```

### 5. Output explanation
The plot shows observations and the matrix reports linear correlations.

### 6. Interpretation / insight
Use the actual output to write a specific observation supported by a number, table, or plot. Note any uncertainty and do not claim causation.

### 7. Best practices
Look for trends, clusters, nonlinear shapes, and outliers.

### 8. Common mistakes
Assuming correlation implies causation.

### 9. Advantages
Combines visual and numeric evidence.

### 10. Limitations
Correlation mainly summarizes linear association.

## Numerical vs categorical

### 1. Concept
Group summaries and box/violin plots compare a numeric distribution across categories.

### 2. Simple example
Compare scores across classrooms.

### 3. Business example
Compare salaries across departments.

### 4. AI/ML relevance
Explore fare by passenger class or age by survival group.

### 5. Output explanation
if "pclass" in df and "fare" in df:
    display(df.groupby("pclass")["fare"].agg(["count","mean","median","std"]))
    sns.boxplot(data=df,x="pclass",y="fare"); plt.title("Fare by passenger class"); plt.show()
if "survived" in df and "age" in df:
    sns.violinplot(data=df,x="survived",y="age",inner="quartile")
    plt.title("Age by survival outcome"); plt.show()

### 6. Interpretation / insight
Small groups and missing values can destabilize results.

### 7. Best practices
Group statistics and plots compare distributions across categories.

### 8. Common mistakes
Check group sizes, overlap, medians, and spread.

### 9. Advantages
Comparing means without considering group sizes or distributions.

### 10. Limitations
Useful group comparisons.

## Categorical vs categorical

### 1. Concept
Cross-tabulation counts category combinations; normalized crosstabs compare proportions.

### 2. Simple example
A table comparing team and selected product.

### 3. Business example
Compare product category with return status.

### 4. AI/ML relevance
Explore passenger class and survival outcome.

```python
Helps inspect target proportions across groups.
```

### 5. Output explanation
if "pclass" in df and "survived" in df:
    display(pd.crosstab(df["pclass"],df["survived"]))
    display(pd.crosstab(df["pclass"],df["survived"],normalize="index").round(3))
    pd.crosstab(df["pclass"],df["survived"],normalize="index").plot(kind="bar",stacked=True)
    plt.title("Survival proportions by class"); plt.ylabel("Proportion"); plt.show()

### 6. Interpretation / insight
Other variables may confound observed relationships.

### 7. Best practices
Counts show group sizes; normalized values show within-group proportions.

### 8. Common mistakes
Compare proportions, not only raw counts.

### 9. Advantages
Assuming group differences are causal.

### 10. Limitations
Interpretable categorical comparison.

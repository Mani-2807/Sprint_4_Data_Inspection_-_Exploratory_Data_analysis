### 1. Concept

### 2. Simple example
Check the dimensions and headings of a spreadsheet before reading it.

### 4. AI/ML relevance
Confirm data types before choosing preprocessing or model inputs.

### 5. Output explanation

### 10. Limitations

## head(), tail(), sample(), info(), describe()

### 1. Concept
`head` shows first rows, `tail` shows last rows, `sample` selects random rows, `info` summarizes structure, and `describe` provides descriptive statistics.

### 2. Simple example
Read the beginning, end, and a random page of a report.

### 3. Business example
Review imported customer or sales data for suspicious formatting.

### 4. AI/ML relevance
Spot obvious issues before creating model features.

```python
display(df.head())
display(df.tail())
display(df.sample(n=min(5, len(df)), random_state=42))
df.info()
display(df.describe(include="all").T)
```

### 5. Output explanation
Tables show sample records, non-null counts, dtypes, and descriptive statistics.

### 6. Interpretation / insight
Use the actual output to write a specific observation supported by a number, table, or plot. Note any uncertainty and do not claim causation.

### 7. Best practices
Compare mean and median, inspect missingness, and look for unexpected ranges.

### 8. Common mistakes
Reviewing only the first few rows and assuming all rows are similar.

### 9. Advantages
Quick first-pass review.

### 10. Limitations
Summaries can hide rare issues and complex patterns.

## nunique() and value_counts()

### 1. Concept
`nunique` counts distinct values; `value_counts` reports category frequencies.

### 2. Simple example
Count the number of unique store names and how often each appears.

### 3. Business example
Find common customer segments or order statuses.

### 4. AI/ML relevance
Identify rare categories and possible target imbalance.

```python
display(df.nunique(dropna=False).to_frame("unique_values"))
for col in df.select_dtypes(exclude=np.number).columns[:6]:
    print("\n", col)
    display(df[col].value_counts(dropna=False).head(15))
```

### 5. Output explanation
Unique counts and frequency tables show category diversity and common labels.

### 6. Interpretation / insight
Use the actual output to write a specific observation supported by a number, table, or plot. Note any uncertainty and do not claim causation.

### 7. Best practices
Investigate high-cardinality columns, rare categories, and null labels.

### 8. Common mistakes
Assuming rare categories are automatically erroneous.

### 9. Advantages
Easy categorical profiling.

### 10. Limitations
Counts alone do not explain why categories occur.

# Notebook 7 — Outlier Analysis
An outlier is an observation unusually far from others under a chosen rule. Do not automatically remove every outlier.

## IQR method

### 1. Concept
The interquartile range is Q3 minus Q1. A common flagging rule marks values below Q1 − 1.5×IQR or above Q3 + 1.5×IQR.

### 2. Simple example
Flag unusually high bills compared with most bills.

### 3. Business example
Investigate very large transactions or unusually long delivery times.

### 4. AI/ML relevance
Extreme observations may affect means, scaling, and some models.

```python
rows=[]
for col in [c for c in ["age","fare","sibsp","parch"] if c in df]:
    s=df[col].dropna(); q1=s.quantile(.25); q3=s.quantile(.75); iqr=q3-q1
    lo=q1-1.5*iqr; hi=q3+1.5*iqr
    rows.append({"column":col,"Q1":q1,"Q3":q3,"IQR":iqr,"lower_fence":lo,"upper_fence":hi,"flagged":int(((s<lo)|(s>hi)).sum())})
display(pd.DataFrame(rows))
```

### 5. Output explanation
The table shows IQR fences and flagged counts for selected columns.

### 6. Interpretation / insight
Use the actual output to write a specific observation supported by a number, table, or plot. Note any uncertainty and do not claim causation.

### 7. Best practices
Review flagged rows against context; a statistical flag is not proof of an error.

### 8. Common mistakes
Automatically deleting every flagged row.

### 9. Advantages
Simple and reasonably robust flagging rule.

### 10. Limitations
Legitimate heavy-tailed values may be flagged.

## Z-score and visual review

### 1. Concept
A z-score measures how many standard deviations a value lies from the mean. It is easiest to interpret for reasonably well-behaved distributions.

### 2. Simple example
Measure a value's distance from the average in standard deviation units.

### 3. Business example
Investigate unusually large amounts.

### 4. AI/ML relevance
Review extreme fares or ages before deciding how to handle them.

```python
Can help identify values needing manual review.
```

### 5. Output explanation
for col in [c for c in ["age","fare"] if c in df]:
    z=(df[col]-df[col].mean())/df[col].std()
    print(col,"abs(z)>3:",int((z.abs()>3).sum()))
    sns.boxplot(x=df[col]); plt.title("Box plot: "+col); plt.show()
if "age" in df and "fare" in df:
    sns.scatterplot(data=df,x="age",y="fare"); plt.title("Age vs fare: joint review"); plt.show()

### 6. Interpretation / insight
Z-scores are sensitive to skew and extreme values.

### 7. Best practices
Counts show values beyond a chosen threshold; plots help inspect them.

### 8. Common mistakes
Decide whether each flag is a data error, valid rare case, or meaningful signal.

### 9. Advantages
Assuming z-score thresholds work equally well for every distribution.

### 10. Limitations
Offers a second flagging method.

# Notebook 4 — Univariate Analysis
Univariate analysis studies one variable at a time.

## Numerical statistics

### 1. Concept
Mean, median, mode, min, max, range, variance, standard deviation, quartiles, percentiles, skewness, and kurtosis describe center, spread, and shape.

### 2. Simple example
Summarize the average, middle value, and spread of exam scores.

### 3. Business example
Understand the distribution of delivery times or transaction amounts.

### 4. AI/ML relevance
Distribution shape can guide later transformation and robust-statistic choices.

```python
num = df.select_dtypes(include=np.number)
display(num.describe().T)
summary = pd.DataFrame({
    "mode": num.mode().iloc[0],
    "range": num.max()-num.min(),
    "variance": num.var(),
    "std_dev": num.std(),
    "skewness": num.skew(),
    "kurtosis": num.kurtosis(),
    "p25": num.quantile(.25),
    "p50": num.quantile(.50),
    "p75": num.quantile(.75),
    "p90": num.quantile(.90)
})
display(summary)
```

### 5. Output explanation
The output summarizes numerical columns; Pandas variance and standard deviation use sample conventions by default.

### 6. Interpretation / insight
Use the actual output to write a specific observation supported by a number, table, or plot. Note any uncertainty and do not claim causation.

### 7. Best practices
Compare mean with median and inspect spread, skewness, and quartiles.

### 8. Common mistakes
Assuming one statistic describes the entire distribution.

### 9. Advantages
Compact numeric profile.

### 10. Limitations
Statistics can be distorted by errors and extreme values.

## Histogram, density plot, and box plot

### 1. Concept
Histograms show counts in bins, density plots estimate distribution shape, and box plots summarize quartiles and potential extremes.

### 2. Simple example
A histogram shows how many passengers fall in each age range.

### 3. Business example
Inspect age or fare distributions before deciding what to investigate.

### 4. AI/ML relevance
Skew and outliers may affect later preprocessing and some models.

```python
for col in [c for c in ["age", "fare", "sibsp", "parch"] if c in df]:
    fig, ax = plt.subplots(1,3,figsize=(13,3))
    sns.histplot(data=df,x=col,ax=ax[0])
    sns.kdeplot(data=df,x=col,fill=True,ax=ax[1])
    sns.boxplot(data=df,x=col,ax=ax[2])
    ax[0].set_title("Histogram: "+col); ax[1].set_title("Density: "+col); ax[2].set_title("Box plot: "+col)
    plt.tight_layout(); plt.show()
```

### 5. Output explanation
Each selected feature has a histogram, density estimate, and box plot.

### 6. Interpretation / insight
Use the actual output to write a specific observation supported by a number, table, or plot. Note any uncertainty and do not claim causation.

### 7. Best practices
Look for asymmetry, long tails, clusters, and potential extremes.

### 8. Common mistakes
Using a density curve for a tiny or discrete dataset without caution.

### 9. Advantages
Multiple views complement statistics.

### 10. Limitations
KDE can mislead for discrete values.

## Categorical frequencies and percentages

### 1. Concept
Frequency counts show category sizes; percentages show each category's share of the dataset.

### 2. Simple example
Count how many orders belong to each store.

### 3. Business example
Compare customer or product category shares.

### 4. AI/ML relevance
Rare or imbalanced categories can affect encoding and evaluation.

```python
for col in df.select_dtypes(exclude=np.number).columns[:6]:
    counts = df[col].value_counts(dropna=False)
    display(pd.DataFrame({"count":counts,"percent":(counts/len(df)*100).round(2)}))
    sns.countplot(data=df,x=col,order=df[col].value_counts().index[:10])
    plt.title("Count plot: "+col); plt.xticks(rotation=30); plt.tight_layout(); plt.show()
```

### 5. Output explanation
Frequency tables and count plots show common and rare labels.

### 6. Interpretation / insight
Use the actual output to write a specific observation supported by a number, table, or plot. Note any uncertainty and do not claim causation.

### 7. Best practices
Check missing categories and whether rare labels are valid.

### 8. Common mistakes
Reading a count plot as a percentage plot.

### 9. Advantages
Easy categorical summary.

### 10. Limitations
Many categories can make charts cluttered.
